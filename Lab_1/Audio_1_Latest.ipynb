## 5.2 Preprocessing (given)

Encode the genre labels, drop the non-numeric `filename` column, and standardize the numeric
features. This part is provided — just run it.

```python
# Label Encoding - encode the categorical genre classes as integers for training
# Blues-0, Classical-1, Country-2, Disco-3, Hip-hop-4, Jazz-5, Metal-6, Pop-7, Reggae-8, Rock-9
class_encod = df.iloc[:, -1]
converter = LabelEncoder()
y = converter.fit_transform(class_encod)

# Drop the filename column - not a numeric feature
df_features = df.drop(labels=["filename", "label"], axis=1, errors='ignore')
if "label" not in df.columns:
    df_features = df.drop(labels=["filename"], axis=1)

# Scale the numeric features
scaler = StandardScaler()
X = scaler.fit_transform(np.array(df_features, dtype=float))

# 70% train / 30% test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print("Train size:", len(y_train), "| Test size:", len(y_test))
```

## 5.3 Baseline models (given)

Two baseline classifiers are provided below: **K-Nearest Neighbors (KNN)** and a **Support Vector
Machine (SVM)**. Run them and note their test-set accuracy — you will need to beat these numbers.

Typical results on this dataset:

| Model | Test accuracy |
|---|---|
| KNN (k=3) | ≈ 0.89 |
| SVM (RBF kernel) | ≈ 0.84 |


```python
# Baseline 1: K-Nearest Neighbors
from sklearn.neighbors import KNeighborsClassifier

knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)

print("KNN training accuracy: {:.3f}".format(knn.score(X_train, y_train)))
print("KNN test accuracy    : {:.3f}".format(knn.score(X_test, y_test)))
print(classification_report(y_test, y_pred_knn))
```

```python
# Baseline 2: Support Vector Machine
from sklearn.svm import SVC

svm_clf = SVC(kernel='rbf')
svm_clf.fit(X_train, y_train)
y_pred_svm = svm_clf.predict(X_test)

print("SVM training accuracy: {:.3f}".format(svm_clf.score(X_train, y_train)))
print("SVM test accuracy    : {:.3f}".format(svm_clf.score(X_test, y_test)))
print(classification_report(y_test, y_pred_svm))
```
